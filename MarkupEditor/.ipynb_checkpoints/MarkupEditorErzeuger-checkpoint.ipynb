```python
#Erzeugt den Code für die replace der alte Markupwerte mit den neuen in XML und XSD Dateien
import pandas as pd
import nbformat as nbf

# CSV-Datei laden
csv_datei = 'markup.csv'
df = pd.read_csv(csv_datei)

# Code-Template vorbereiten
code_template = """
# XML-Datei laden
with open('input.xml', 'r', encoding='utf-8') as file:
    data = file.read()

# Änderungen durchführen
data = data.replace("#A-Wert#", "#B-Wert#")

# Geänderte Inhalte in derselben Datei speichern
with open('input.xml', 'w', encoding='utf-8') as file:
    file.write(data)
"""

# Werte aus der CSV vergleichen und Code anpassen
replace_statements = ""
for index, row in df.iterrows():
    a_wert = str(row['A'])
    b_wert = str(row['B'])
    if a_wert != b_wert:
        replace_statements += f"data = data.replace('\"{a_wert}\"', '\"{b_wert}\"')\n"

# Den Code mit den generierten `.replace()`-Befehlen füllen
final_code = code_template.replace('data = data.replace("#A-Wert#", "#B-Wert#")', replace_statements)

# Notebook erstellen
nb = nbf.v4.new_notebook()
nb.cells.append(nbf.v4.new_code_cell(final_code))

# Datei speichern
with open('MarkupEditor.ipynb', 'w', encoding='utf-8') as f:
    nbf.write(nb, f)

print("Die Datei 'MarkupEditor.ipynb' wurde erfolgreich erstellt.")
```
